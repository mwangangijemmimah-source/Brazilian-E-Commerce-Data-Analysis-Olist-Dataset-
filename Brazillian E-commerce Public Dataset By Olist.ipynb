```python
#Total count  of missing values and % of Total orders
import pandas as pd
orders = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_orders_dataset.csv")
missing_count = orders['order_delivered_customer_date'].isnull().sum()
total_orders = len(orders)
percentage_missing = (missing_count / total_orders) * 100
print(f"Total Missing Delivery Dates: {missing_count}")
print(f"Percentage of Total Orders: {percentage_missing:.2f}%")
```

```python
##English product category with highest average product weight & the average weight
import pandas as pd
products =pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_products_dataset.csv")
translation = pd.read_csv(r"C:\Users\LENOVO\Desktop\product_category_name_translation.csv")

products_merged = pd.merge(products, translation, on='product_category_name', how='left')
avg_weights = products_merged.groupby('product_category_name_english')['product_weight_g'].mean()

heaviest_category = avg_weights.idxmax()
max_weight = avg_weights.max()

print(f"Heaviest Category: {heaviest_category}")
print(f"Average Weight: {max_weight:.2f} grams")
```

```python
##Maximum number of installments recorded Payment type ccounts fot the highest total value when installments exceed 5
import pandas as pd
payments = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_order_payments_dataset.csv")

max_installments = payments['payment_installments'].max()
high_installments = payments[payments['payment_installments'] > 5]

value_by_type = high_installments.groupby('payment_type')['payment_value'].sum()

top_payment_type = value_by_type.idxmax()
top_total_value = value_by_type.max()

print(f"Maximum Installments: {max_installments}")
print(f"Top Payment Type (>5 installments): {top_payment_type}")
print(f"Total Value: {top_total_value:.2f}")
```

```python
##Number of order items fulfilled by the seller located in the same state as the purchasing customer
import pandas as pd
sellers = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_sellers_dataset.csv")
orders = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_orders_dataset.csv")
order_items = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_order_items_dataset.csv")
customers = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_customers_dataset.csv")

merged = order_items.merge(sellers, on='seller_id') \
    .merge(orders, on='order_id') \
        .merge(customers, on='customer_id')
        
        
same_state_count = (merged['seller_state'] == merged['customer_state']).sum()


total_items = len(merged)
percentage = (same_state_count / total_items) * 100

print(f"Items fulfilled in the same state: {same_state_count}")
print(f"Total items analyzed: {total_items}")
print(f"Percentage of 'Local' fulfillment: {percentage:.2f}%")
```

```python
##Mean Review Score for orders that arrived on time vs those that arrived late
import pandas as pd
orders = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_orders_dataset.csv")
reviews = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_order_reviews_dataset.csv")


orders['order_delivered_customer_date'] = pd.to_datetime(orders['order_delivered_customer_date'])
orders['order_estimated_delivery_date'] = pd.to_datetime(orders['order_estimated_delivery_date'])


delivered = orders.dropna(subset=['order_delivered_customer_date']).copy()


delivered['is_late'] = delivered['order_delivered_customer_date'] > delivered['order_estimated_delivery_date']


merged = pd.merge(delivered, reviews, on='order_id')


results = merged.groupby('is_late')['review_score'].mean()

print(f"Average Review Score (On Time): {results[False]:.2f}")
print(f"Average Review Score (Late): {results[True]:.2f}")

```

```python
##Total number of items sold per english category and visualize the top 10 categories
import pandas as pd
import matplotlib.pyplot as plt

products = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_products_dataset.csv")
order_items = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_order_items_dataset.csv")
translation = pd.read_csv(r"C:\Users\LENOVO\Desktop\product_category_name_translation.csv")

merged = order_items.merge(products, on='product_id') \
   .merge(translation, on='product_category_name')
   

category_counts = merged['product_category_name_english'].value_counts()


top_10 = category_counts.head(10)


top_10.sort_values().plot(kind='barh', figsize=(10, 6), color='skyblue')

plt.title('Top 10 Product Categories by Items Sold')
plt.xlabel('Number of Items Sold')
plt.ylabel('Product Category')
plt.show()
```

```python
# Total Transaction Value & create a pie chart showing % Market share of each payment method
import pandas as pd
payments = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_order_payments_dataset.csv")
payment_shares = payments.groupby('payment_type')['payment_value'].sum()
payment_shares.plot(
    kind='pie', 
    figsize=(8, 8), 
    autopct='%1.1f%%', 
    startangle=140,
    title='Market Share of Payment Methods'
)

plt.ylabel('') 
plt.show()
```

```python
##Time Taken to Approve  each order & plot an histogram
import pandas as pd
import matplotlib.pyplot as plt
orders = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_orders_dataset.csv")

orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])
orders['order_approved_at'] = pd.to_datetime(orders['order_approved_at'])

# Approval time in hours

orders['approval_time_hours'] = (orders['order_approved_at'] - orders['order_purchase_timestamp']).dt.total_seconds() / 3600


filtered_data = orders[orders['approval_time_hours'].notnull() & (orders['approval_time_hours'] <= 72)]


filtered_data['approval_time_hours'].plot(
    kind='hist', 
    bins=50, 
    figsize=(10, 6), 
    color='salmon', 
    edgecolor='black',
    title='Distribution of Order Approval Times'
)

plt.xlabel('Hours from Purchase to Approval')
plt.ylabel('Number of Orders')
plt.show()
```

```python
##Total monthly revenue, visualize revenue growth trend over time
import pandas as pd
import matplotlib.pyplot as plt
orders = pd.read_csv(r"C:\Users\LENOVO\Desktop\olist_orders_dataset.csv")


orders['order_purchase_timestamp'] = pd.to_datetime(orders['order_purchase_timestamp'])


merged = pd.merge(orders, order_items, on='order_id')


merged.set_index('order_purchase_timestamp', inplace=True)
monthly_revenue = merged['price'].resample('ME').sum()


monthly_revenue.plot(
    kind='line', 
    figsize=(12, 6), 
    marker='o', 
    color='teal', 
    title='Monthly Revenue Growth Trend'
)

plt.xlabel('Month')
plt.ylabel('Total Revenue (BRL)')
plt.grid(True)
plt.show()
```
